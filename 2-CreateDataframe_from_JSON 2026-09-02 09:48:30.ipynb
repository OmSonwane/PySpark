#Creating a Dataframe from JSON

```python
# JSON file path-     /Volumes/workspace/default/sparkhandsonfiles/movies.json
# Nested-JSON file path-    /Volumes/workspace/default/sparkhandsonfiles/nested_json.json
```

```python
df_json = spark.read.format('json')\
                    .option('mode','PERMISSIVE')\
                    .option('inferSchema',True)\
                    .load(r'/Volumes/workspace/default/sparkhandsonfiles/movies.json')
```

```python
df_json.display()
```

# Creating a Dataframe from Nested JSON

```python
df_nestedjson = spark.read.format('json')\
                            .option('mode','PERMISSIVE')\
                            .option('inferSchema',True)\
                            .option('multiLine',True)\
                            .load(r'/Volumes/workspace/default/sparkhandsonfiles/nested_json.json')
```

```python
df_nestedjson.show()
```

```python
df_nestedjson.printSchema()
```

```python
df_nestedjson.explain()
```
