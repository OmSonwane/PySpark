## Different Ways to refer a column

```python
df = spark.read.format('csv')\
                .option("inferSchema",True)\
                .option("mode","PERMISSIVE")\
                .option("header",True)\
                .load(r'/Volumes/workspace/default/sparkhandsonfiles/employee.csv')
```

```python
df.show()
```

```python
df.printSchema()
```

### Way 1 : using column names directly as string

```python
df.select('eid','ename').show()
```

```python
df.filter("salary">50000)
```

```python
### We cannot perform transformation on colummns that are referred as a String. Solutions use other ways to refer a column. 
```

### Way 2 : using col() function

```python
from pyspark.sql.functions import col
```

```python
df.select(col('eid'),col('ename')).show()
```

```python
df.filter(col('salary')>100000).show()
```

### Way 3 : using dataFrame attribute-style access

```python
df.select(df.ename,df.salary,df.dept).show()
```

```python
df.filter(df.dept == 'HR').show()
```

### Way 4 : using selectExpr for SQL-like expressions

```python
df.selectExpr("ename as empName","dept as empDepartment").show()
```

```python
df.selectExpr("MAX(salary) AS maxSalary").show()
```

### Way 5 : using SQL directly by registering a temporary view

```python
# You can't perform SQL query directly on Dataframe, instead create a temporary view from dataframe and then apply query on that view
```

```python
df.createOrReplaceTempView("view_employee")
```

```python
spark.sql("SELECT * FROM view_employee").show()
```

```python
spark.sql("SELECT ename,eid,dept FROM view_employee WHERE salary>100000").show()
```

```python
df.selectExpr("SELECT * FROM view_employee").show()
```
