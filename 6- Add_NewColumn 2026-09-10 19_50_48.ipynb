# Adding New Column

```python
# /Volumes/workspace/default/sparkhandsonfiles/employee.csv
```

```python
df = spark.read.format('csv')\
                .option('inferSchema',True)\
                .option('header',True)\
                .option('mode','PERMISSIVE')\
                .load(r'/Volumes/workspace/default/sparkhandsonfiles/employee.csv')
```

```python
df.show()
```

```python
df.printSchema()
```

## way 1: using select()

```python
df.select("*",(df.salary*1.10).alias('incremeted_salary')).show()
```

## way 2: using selectExpr()

```python
df.selectExpr("*","salary*1.10 AS incremented_sal").show()
```

## way 3: using select() and expr()

```python
from pyspark.sql.functions import expr
```

```python
df.select("*",expr("salary*1.10 AS incr_sal")).show()
```

## way 4: using withColumn()

```python
df.withColumn('incr_salary',df.salary*1.10).show()
```

```python
df.show()
```

```python
# We are not storing the result into df that's why we're not getting new columns in that dataframe...Dataframes are immutable!!
df1 = df # Not it will store the result into df1
```

```python
df1.show()
```

```python
df1 = df1.select("*",df1.salary*2.0)
```

```python
df1.show()
```

```python
df2 = df1.select('eid','ename','dept','date_of_joining',df.salary*2.0).show()
```

```python
# Now, we are just transforming df1 and storing the result in df2, which is fine....But, what if we store it in df1 only?
```

```python
df1 = df1.select('eid','ename','dept','date_of_joining',(df1.salary*12).alias("Package")).show()
```

```python
# As far we now that dataframes are immutable. Then how df1 has been changed? So, the answer is df1 is not changed, it has just changed the referencing to new result. df1 was poiting to result of cell 18, but now it has changed it pointer to result of cell 22...And due to this we won't be able to access result of cell 18 with the help of df1's pointer. In fact, df1's pointer will lead us to result of cell 22. To re-access result of cell 18 with df1's pointer we will have to re-run the logic of df1's creation and then we can access it later on. hence, dataframe is not changed, we have only changed the pointer. We are not updating the dataframe but updating the reference. We have re-declared the df varibale only, not anythig else. in programming context, Immutablity does not mean re-declaration is not possible, but re-definition is. Like Python, every definitoon is re-declaration in PySpark.
```
