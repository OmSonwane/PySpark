# Creating Dataframe from Binary File Formats

```python
# ORC file path- /Volumes/workspace/default/sparkhandsonfiles/cr.orc
# AVRO file path- /Volumes/workspace/default/sparkhandsonfiles/twitter.avro
# PARQUET file path- /Volumes/workspace/default/sparkhandsonfiles/movies.parquet
```

### Create DF from ORC 

```python
df_orc = spark.read.format('orc')\
                    .load(r'/Volumes/workspace/default/sparkhandsonfiles/cr.orc')
```

```python
df_orc.show()
```

```python
df_orc.printSchema()
```

```python
# here we did not given any schema or used inferSchema because Text Based File Formats do not need to mention schema. That is why they are fast to process
```

### Creating DF from AVRO

```python
df_avro = spark.read.format('avro')\
                    .load(r'/Volumes/workspace/default/sparkhandsonfiles/twitter.avro')
```

```python
df_avro.display(Truncate=False)
```

```python
df_avro.printSchema()
```

### Creating DF from PARQUET

```python
df_parquet = spark.read.format('parquet')\
                        .load(r'/Volumes/workspace/default/sparkhandsonfiles/movies.parquet')
```

```python
df_parquet.show()
```

```python
df_parquet.printSchema()
```

```python
### The issue is that in Databricks, the default format for spark.read.load() is Delta, not Parquet. When you don't specify a format, Spark tries to read the file as a Delta table, which requires a transaction log (_delta_log directory). Your movies.parquet file is a plain Parquet file without Delta metadata, so it fails. In cell 13, you correctly specified .format('parquet'), which worked. Cell 17 omits the format, triggering the error. Previously there was parquet format as a default and that is why this code used to get run successfully, but now Spark has change default file format to Delta. Delta file is actually a folder containing Parquet files plus a _delta_log
```
