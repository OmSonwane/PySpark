# Create DF From CSV File

### Method 1: inferschema = True 

```python
# CSV file path
/Volumes/workspace/default/sparkhandsonfiles/employee.txt

#Pipe separated file path
/Volumes/workspace/default/sparkhandsonfiles/users_pipe_separated.txt
```

```python
df = spark.read.format('csv').\
                option('header', True).\
                option('inferschema', True).\
                option('mode','PERMISSIVE').\
                load('/Volumes/workspace/default/sparkhandsonfiles/employee.txt')       
```

```python
df.show()
```

```python
df.printSchema()
```

### Method 2: Explicit Schema by StructType, StructField

```python
from pyspark.sql.types import *
```

```python
mySchema = StructType([
    StructField("empID",IntegerType(),False),
    StructField("empName",StringType(),False),
    StructField("empDept",StringType(),True),
    StructField("empSal",FloatType(),True),
    StructField("DateOfJoining",DateType(),True)
])
```

```python
mySchema_df = spark.read.format("csv").\
                        option("header",True).\
                        schema(mySchema).\
                        option('mode','PERMISSIVE').\
                        load("/Volumes/workspace/default/sparkhandsonfiles/employee.txt")
```

```python
mySchema_df.printSchema()
```

### Method 3: Explicit Schema by DDL

```python
mySchema2 = "empID INT, empName STRING, empDept STRING, empSal DOUBLE, joiningDate DATE"
```

```python
mySchema2_df = spark.read.format('csv').\
                            option('header',True).\
                            option('mode','DROPMALFORMED').\
                            schema(mySchema2).\
                            load('/Volumes/workspace/default/sparkhandsonfiles/employee.txt')    
```

```python
mySchema2_df.show()
```

```python
mySchema2_df.printSchema()
```

### Creating Dataframe from Pipe Delimited File

```python
df_pipe = spark.read.format('csv').\
                    option('mode','PERMISSIVE').\
                    option('header',True).\
                    option('delimiter','|').\
                    option('inferSchema',True).\
                    load(r'/Volumes/workspace/default/sparkhandsonfiles/users_pipe_separated.txt')
```

```python
df_pipe.show()
```

```python
df_pipe.printSchema()
```
